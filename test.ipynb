```python
from typing import List, Optional

import pandas as pd
from sqlmodel import Field, SQLModel


class Member(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    weight: float = Field(default=0.0)
    gender: str = Field(default="")
    club: str = Field(default="")


# Create a few users for example
member = [
    Member(id=0, weight=80.0, gender="M", club="Club A"),
    Member(id=1, weight=70.0, gender="M", club="Club A"),
    Member(id=2, weight=60.0, gender="M", club="Club B"),
    Member(id=3, weight=50.0, gender="F", club="Club C"),
    Member(id=4, weight=55.0, gender="F", club="Club B"),
]


def sqmodel_to_df(objs: List[SQLModel]) -> pd.DataFrame:
    """Convert a SQLModel objects into a pandas DataFrame."""
    records = [i.dict() for i in objs]
    df = pd.DataFrame.from_records(records)
    return df

df = sqmodel_to_df(member)
print(df)

#      id             email  password    name
# 0  None     sam@gmail.com      1234     sam
# 1  None    jake@gmail.com      abcd    jake
# 2  None  olivia@gmail.com  password  Olivia
```

```python
sorted_df = df.sort_values(by="club", ascending=False)
print(sorted_df.index)
```
